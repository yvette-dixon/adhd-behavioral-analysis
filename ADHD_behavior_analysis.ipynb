```python
# Import the pandas library
import pandas as pd

# Loading the dataset (CSV file) into a pandas DataFrame
df = pd.read_csv("adhd_data.csv")
```

```python
# Confirming if the dataset is loaded correctly by asking for a sample
df.head()
```

```python
# Checking the shape of the dataset to confirm if all data uploaded
df.shape
```

```python
# Check for missing values in each column to ensure data quality
df.isnull().sum()
```

Result: All columns returned 0, meaning there are no missing values in the dataset. Therefore, no action is needed for handling missing data.

```python
# Running to view duplicated rows in the dataset, if any
df.duplicated().sum()
```

Result: 0 duplicate rows were found in the dataset. Therefore, no rows needed to be removed

```python
# Standardize column names by converting them to lowercase and replacing hyphens with underscores
df.columns = df.columns.str.lower().str.replace("-", "_")

# Display updated column names
df.columns
```

```python
# Converting selected columns to string and standardize to lowercase

df['family_history'] = df['family_history'].astype(str).str.lower()
df['educational_level'] = df['educational_level'].astype(str).str.lower()
df['diagnosis_class'] = df['diagnosis_class'].astype(str).str.lower()

# Note: 'gender' is numeric (1 and 2), so no text conversion is applied
```

The "numbers" column is not saved as a text string

```python
# Convert selected columns to string type and standardize text to lowercase
# This ensures consistency in categorical data (ex. "Male" vs "male")
df['family_history'] = df['family_history'].astype(str).str.lower()
df['educational_level'] = df['educational_level'].astype(str).str.lower()
df['diagnosis_class'] = df['diagnosis_class'].astype(str).str.lower()
```

Note: Gender column contains numeric values (1 and 2), so it was not converted to lowercase

```python
# Save the cleaned dataset
df.to_csv("cleaned_adhd_data.csv", index=False)
```

**Exploratory Data Analysis**

```python
# Using describe to summarize the data using mean, min, max, standard deviation, etc.
df.describe()
```

```python
# Group data by ADHD diagnosis and calculate the average of numeric columns
# numeric_only=True is used to avoid errors from text columns
df.groupby('diagnosis_class').mean(numeric_only=True)
```

```python
df.groupby('diagnosis_class').mean(numeric_only=True)
```

```python
# Filtering dataset to only include individuals with higher ADHD levels to focus on those with stronger ADHD symptoms
high_adhd = df[df['diagnosis_class'].isin(['2', '3'])]

high_adhd.head()
```

```python
import matplotlib.pyplot as plt

# Bar chart showing average focus score by ADHD diagnosis
df.groupby('diagnosis_class')['focus_score_video'].mean().plot(kind='bar')

# Add title and labels
plt.title("Average Focus Score by ADHD Diagnosis")
plt.xlabel("ADHD Diagnosis Class")
plt.ylabel("Average Focus Score")

plt.show()
```

```python
# Histogram of daily phone usage
# This shows how phone usage is distributed across individuals
df['daily_phone_usage_hours'].plot(kind='hist', bins=20)

plt.title("Distribution of Daily Phone Usage")
plt.xlabel("Hours of Phone Usage")
plt.ylabel("Frequency")

plt.show()
```

```python
# Scatter plot showing relationship between phone usage and focus score
# This helps examine if screen time is related to attention levels
plt.scatter(df['daily_phone_usage_hours'], df['focus_score_video'])

# Add title and labels
plt.title("Phone Usage vs Focus Score")
plt.xlabel("Daily Phone Usage (hours)")
plt.ylabel("Focus Score")

plt.show()
```

```python
# Computing correlation matrix for numeric variables to show how strong variables relate to each other
correlation_matrix = df.corr(numeric_only=True)

# Display the correlation matrix
correlation_matrix
```

```python
# Extract correlation values specifically for ADHD diagnosis
adhd_corr = correlation_matrix['diagnosis_class'].sort_values(ascending=False)

adhd_corr
```

```python
# Make sure diagnosis_class is numeric
df['diagnosis_class'] = pd.to_numeric(df['diagnosis_class'], errors='coerce')

# Recompute correlation matrix
correlation_matrix = df.corr(numeric_only=True)
```
